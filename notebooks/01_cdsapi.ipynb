```python
import cartopy.crs as ccrs
import cartopy.feature as cfeature
```

```python
sea_ice_conc = ds['ice_conc']
```

```python
sea_ice_conc
```

```python
map_projection = ccrs.NorthPolarStereo(central_longitude=0)
```

```python
extent = [-180, 180, ds['lat'].min().item(), 90]
```

```python
extent
```

```python
ds
```

```python
fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(1, 1, 1, projection=map_projection)
```

```python
c = ax.pcolormesh(
    ds['lon'], ds['lat'], sea_ice_conc.isel(time=0),
    transform=ccrs.PlateCarree(),
    cmap='Blues_r',
    shading='auto',
    vmin=0, vmax=1
)
```

```python
fig = plt.figure(figsize=(10,10))
ax = fig.add_subplot(1, 1, 1, projection=map_projection)
```

```python
plt.colorbar(c, ax=ax, orientation='vertical', label='Sea Ice Concentration (%)')
ax.set_title('Sea Ice Concentration (Without Cartopy)')
ax.set_xlabel('Longitude')
ax.set_ylabel('Latitude')
ax.set_aspect('auto') # You might play with 'equal' but it often looks weird
plt.show()
```
